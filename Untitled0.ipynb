```python
# 1. 原始輸入資料（字串格式）
raw_meal_cost = "$1,200"
raw_people_count = "4"
coupon = 100
budget_per_person = 300

# 2. 字串處理與型別轉換
# 移除美元符號與千分位逗號，並將字串轉為整數
cleaned_cost_str = raw_meal_cost.replace("$", "").replace(",", "")
original_cost = int(cleaned_cost_str)

people_count = int(raw_people_count)

# 3. 費用計算步驟（分開寫以保持清晰）
# 規則一：餐廳按原始餐費收取 10% 服務費
service_fee = original_cost * 0.1

# 規則二：計算服務費後，再扣除折價券，得出應付總額
total_amount = original_cost + service_fee - coupon

# 規則三：所有人平均分攤應付總額
per_person_cost = total_amount / people_count

# 規則四：檢查是否超過每人 300 元預算（回傳布林值 True 或 False）
is_over_budget = per_person_cost > budget_per_person

# 4. 基本輸出與費用摘要
print("=== 聚餐費用摘要 ===")
print(f"原始餐費：NT$ {original_cost}")
print(f"服務費 (10%)：NT$ {service_fee}")
print(f"折價券折抵：-NT$ {coupon}")
print(f"聚餐應付總額：NT$ {total_amount}")
print(f"參與人數：{people_count} 人")
print(f"每人平均分攤：NT$ {per_person_cost}")
print(f"是否超過每人 300 元預算？ {is_over_budget}")

```
